```python
import numpy as np
from integrator import ComplexIntegrator


c = 10-100j
def foo(x):
    return 1/(x-c)

def param(x):
    return (x[:,0] - 0.5) * 10, 10

numeric_int = ComplexIntegrator(1).integrate(foo,param,n_epochs=100, samples_per_epoch=10_000)
algebraic_int = np.log((5-c)/(-5-c))

print(numeric_int)
print(algebraic_int)
```
