# Log integral

```python
import numpy as np
from integrator import ComplexIntegrator
import cmath

def get_foo(c):
    def foo(x):
        return 1/(x-c)
    return foo

def get_param(lim):
    def param(x):
        z = (2*x[:,0] - 1) * lim      
        jac = 2 * lim                 
        return z, np.full_like(z,jac)
    return param


def integrate_twice(c, lim):
    ci = ComplexIntegrator(1)
    print(f'c = {c}')
    ci_val = ci.integrate(get_foo(c), get_param(lim), 500,2000)
    ana = cmath.log((lim-c)/(-lim-c))
    print(ci_val)
    print(f'{ana:.6f}')
    

integrate_twice(0+0.1j,1)
    
```
